## Important Note on Missing Values

The dataset originally contains missing values represented as the string `'?'`, which is **not recognized by pandas** as a true missing value (`NaN`).

To fix this:
- We scan each column for `'?'`
- Replace all `'?'` with `np.nan`
- Then use `df.isnull().sum()` to correctly identify missing values

This step ensures that the dataset is properly prepared for cleaning and modeling.

```python
# Check how many '?' are present in each column
for col in df.columns:
    print(f"{col} -> {(df[col] == '?').sum()} '?' values")
```

```python
# Replace '?' with actual missing values (NaN)
df.replace('?', np.nan, inplace=True)
```

```python
# Now these will show True where missing values are present
print(df.isnull().sum())
```

## Handling Missing Values

We observed missing values in the following columns:
- `ca`: 4 missing entries
- `thal`: 2 missing entries

Both are categorical in nature. Instead of dropping these rows, we imputed the missing values with the **mode** (most frequent value) of each column.

This helps preserve the dataset size while maintaining data integrity.


```python
# Fill with mode (msot frequent values)
df['ca'].fillna(df['ca'].mode()[0], inplace=True)
df['thal'].fillna(df['thal'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

## Data Type Conversion

After replacing the missing values, we observed that the columns `ca` and `thal` were stored as **object** type due to the presence of `'?'` values earlier.

We now convert them to proper **numeric** types (`int` or `float`) using `pd.to_numeric()` to ensure they are treated correctly during model training.


```python
# Convert 'ca' and 'thal' to numeric

df['ca'] = pd.to_numeric(df['ca'])
df['thal'] = pd.to_numeric(df['thal'])

print(df.dtypes[['ca', 'thal']])
```

```python
# Check unique values and their counts in 'target'
df['target'].value_counts()
```

### Target Variable: Class Distribution 

The `target` column in the original UCI Heart Disease dataset contains **5 classes**:

- `0`: No Heart Disease  
- `1–4`: Different levels of Heart Disease severity  

**Class Distribution:**

- 0 → 164 instances  
- 1 → 55 instances  
- 2 → 36 instances  
- 3 → 35 instances  
- 4 → 13 instances  

For simplicity and better interpretability, we will convert this **multi-class target** into a **binary classification problem**:

- `0` → **No Heart Disease**
- `1–4` → **Heart Disease** (will be converted to `1`)

```python
# Convert multi-class target to binary target
df['target'] = df['target'].apply(lambda x: 1 if x > 0 else 0)
```

```python
import joblib
import pandas as pd
model = joblib.load('/Heart_Disease_Predictor/models/best_model.pkl')
scaler = joblib.load('/Heart_Disease_Predictor/models/scaler.pkl')
```

```python
# Define feature names
feature_names = ['age', 'sex', 'cp', 'thalach', 'exang', 'oldpeak', 'ca', 'thal']

# Example input (you can update this or take dynamic input)
input_data = [[54, 1, 1, 135, 0, 1.4, 1, 1]]

# Create DataFrame
new_data = pd.DataFrame(input_data, columns=feature_names)

# Scale the input
new_data_scaled = scaler.transform(new_data)

# Predict class and probability
prediction = model.predict(new_data_scaled)
probability = model.predict_proba(new_data_scaled)[0]

# Result label
result = "Heart Disease" if prediction[0] == 1 else "No Heart Disease"

# Output
print("Prediction:", prediction[0])
print("Result    :", result)
print(f"Probability of No Heart Disease [0]: {probability[0]:.2f}")
print(f"Probability of Heart Disease   [1]: {probability[1]:.2f}")
```
